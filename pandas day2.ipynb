```python
import pandas as pd

data = {
    "Date": ["2024-01-01","2024-02-01","2024-01-02","2024-02-02"],
    "Region": ["East","West","East","West"],
    "Product": ["A","A","B","B"],
    "Sales": [100, 150, 200, 250],
    "Quantity": [2, 3, 4, 5]
}

Df = pd.DataFrame(data)
Df["Date"] = pd.to_datetime(Df["Date"])
Df
```

```python
pivot_Df = pd.pivot(index = 'Date', columns='Product', values = 'Sales', data=Df)
pivot_Df
```

```python
pivot_table_df = pd.pivot_table(
    Df,
    index="Date",
    columns="Region",
    values="Sales",
    aggfunc=["sum","mean"]
)
pivot_table_df
```

```python
Df.groupby("Region").agg({
    "Sales": "sum",
    "Quantity": "mean"
})

```

```python
pivot_table_df.reset_index().columns
```

```python
pivot_df = pivot_table_df.reset_index()
pivot_df.columns = [
    col[0] if col[1] == "" else f"{col[0]}_{col[1]}"
    for col in pivot_df.columns
]
```

```python
melted = pd.melt(
    pivot_df,
    id_vars="Date",
    var_name="Metric_Region",
    value_name="Sales"
)
melted
```
